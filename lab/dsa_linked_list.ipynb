```python
def add_node_to_last(head: Node = None, new_node: Node = None) -> Node:
    if head is None:
        head = new_node
    
    last: Node = head
    while last.pointer is not None:
        last = last.pointer
    
    last.pointer = new_node
    return head
```

```python
def add_node_to_specific_position(head: Node = Node, new_node: Node = None, position: int = 0):
    if not new_node:
        return None
    if not head:
        head = new_node
        return head
    if position == 0:
        add_node_at_first(head, new_node)
        return head
```

```python
# Complete

class Node:
    def __init__(self, value: str|int|bool, pointer: "Node" = None) -> None:
        self.value = value
        self.pointer = pointer

class LinkedList:
    def __init__(self, node: Node = None):
        self.head = node
        self.len = 1 if self.head else 0

    def print_linked_list(self) -> None:
        current = self.head
        while current:
            print(current.value)
            current = current.pointer
        print(None)

    def get_len(self) -> int:
        return self.len

    def add_node_to_first(self, node: Node):
        node.pointer = self.head
        self.head = node
        self.len += 1
        return self.head

    def add_node_to_last(self, new_node: Node) -> Node:
        if new_node is None:
            raise ValueError("new_node cannot be None")
        if new_node.pointer is not None:
            raise ValueError("new_node must be a detached node")

        if self.head is None:
            self.head = new_node
            self.len += 1
            return self.head
        last = self.head
        while last.pointer is not None:
            last = last.pointer
        last.pointer = new_node
        self.len += 1
        return self.head

    def remove_first(self) -> None:
        if self.head is None:
            return
        self.head = self.head.pointer
        self.len -= 1

    def remove_last(self) -> None:
        if self.head is None:
            return

        if self.head.pointer is None:
            self.head = None
            self.len -= 1
            return

        current = self.head
        while current.pointer.pointer is not None:
            current = current.pointer

        current.pointer = None
        self.len -= 1

    def search(self, value: any) -> bool:
        current = self.head
        while current:
            if current.value == value:
                return True
            current = current.pointer
        return False

    def reverse(self) -> None:
        if self.head is None:
            return

        prev = None
        current = self.head
        while current:
            next = current.pointer
            current.pointer = prev
            prev = current
            current = next
        self.head = prev
        return self.head

    def get_by_index(self, index) -> bool:
        if index > self.len -1 or index < 0:
            raise ValueError("index is out of range")
        current = self.head
        for i in range(index):
            current = current.pointer
        return current.value
```

```python
def find_middle(linked_list) -> Node | None:
        """
        Finds the middle node of the linked list using the fast and slow pointer method.
        Returns the second middle node if the list has an even number of nodes.
        Source: https://www.youtube.com/watch?v=A2_ldqM4QcY
        """
        slow = linked_list.head
        fast = linked_list.head

        # The loop condition handles both odd and even length lists
        while fast is not None and fast.pointer is not None:
            slow = slow.pointer
            fast = fast.pointer.pointer

        return slow # The slow pointer is now at the middle node

find_middle(linked_list).value
```

```python
def move_three_steps(linked_list):
    slow = fast = linked_list.head

    while (
        fast is not None
        and fast.pointer is not None
        and fast.pointer.pointer is not None
    ):
        slow = slow.pointer
        fast = fast.pointer.pointer.pointer

    return slow

move_three_steps(linked_list).value
```

```python
import snoop
import sys
snoop.Config(color=True, out=sys.stdout, overwrite=True)

@snoop
def add_int(num: list[int], k: int) -> list[int]:
    arr_value = 0
    for i in range(len(num) - 1, -1, -1):
        arr_value += num[len(num) - i - 1] * (10 ** i)
    arr_value += k
    return list(map(lambda x: int(x), list(str(arr_value))))
add_int([2,7,4], 181)
```

```python
@snoop
def add_int_v2(num: list[int], k:int) -> list[int]:
    """
    https://www.youtube.com/watch?v=eBTZQt1TWfk
    """
    num.reverse()
    i = 0
    while k:
        digit = k % 10
        if i < len(num):
            num[i] += digit
        else:
            num.append(digit)
    
        carry = num[i] // 10
        num[i] = num[i] % 10
        
        k = k // 10
        k += carry
        i += 1
    num.reverse()
    
    return num

add_int_v2([2,7,4], 181)
```
